## 3. Basic test of the Team class

This creates a Team object, adds two Driver objects, and checks the team's total points.

```python
test_team = Team("FERRARI")

test_team.add_driver(Driver("Carlos Sainz", 200))
test_team.add_driver(Driver("Charles Leclerc", 206))

print(test_team)
print("Total points:", test_team.get_total_points())
```

## 4. Load the F1 data from the CSV file

Each row contains a driver's name, team, and points.

A dictionary is used so that each team is only created once.

```python
teams = {}

with open("f1_points.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        driver_name = row["Driver"]
        team_name = row["Team"]
        points = int(row["Points"])

        if team_name not in teams:
            teams[team_name] = Team(team_name)

        driver = Driver(driver_name, points)
        teams[team_name].add_driver(driver)
```

## 5. Check the loaded data

The lab expects 10 Team objects.

This cell shows:
- the number of teams created
- the team names

```python
print("Number of teams:", len(teams))
print("Teams loaded:")

for team_name in teams:
    print(team_name)
```

## 6. Show each team and its drivers

This demonstrates that each Team object contains Driver objects.

```python
for team in teams.values():
    print(team)
```

## 7. Sort the teams by total points

The Team class contains a `__lt__` method.

That allows Python's `sorted()` function to compare Team objects using their total points.
The teams will be sorted from lowest total points to highest total points.

```python
sorted_teams = sorted(teams.values())

for team in sorted_teams:
    print(team)
```

## 8. Test Cases

The following cells demonstrate normal and special cases.

### Test 1: Normal case

A team with two drivers should return the sum of both drivers' points.

Expected result: `175`

```python
normal_team = Team("TEST TEAM")

normal_team.add_driver(Driver("Driver One", 100))
normal_team.add_driver(Driver("Driver Two", 75))

print("Expected: 175")
print("Actual:", normal_team.get_total_points())
```

### Test 2: Special case

A team with no drivers should have a total of `0` points.

Expected result: `0`

```python
empty_team = Team("EMPTY TEAM")

print("Expected: 0")
print("Actual:", empty_team.get_total_points())
```

### Test 3: Team comparison

The `__lt__` method should return `True` when the first team has fewer points than the second.

Expected result: `True`

```python
team_a = Team("TEAM A")
team_b = Team("TEAM B")

team_a.add_driver(Driver("Driver A", 50))
team_b.add_driver(Driver("Driver B", 100))

print("Expected: True")
print("Actual:", team_a < team_b)
```
